##### Reasoning And Decision For The Choice Of Random Forest Classifier Algorithm

```python
#    THE DATASET

# 1. The Telco Customer Churn Dataset is a combination of different categories of features (X) data values with 21 columns, 
#....and a target column (Y).

# 2. The break-down of the different categories of the data values and their data types shows some of the feature columns 
#....with continuous numerical values, combined with other categorical columns with simple binary values and multiple
#....categories of texts, as well as simple and more complex binary values. 

# 3  The dataset are also labelled with inputs resulting in known outputs, as the predictor columns for a supervised classification
#... learning.

# 4  The labelled datasets show the relationship between inputs and outputs (independent i.e. features (X) and dependent i.e.
#....target (Y) values, for predicting a categorical values of target (Y) and labels of unseen and new data, with the Decision
#.... Tree Classifier machine learning algorithm, wherein inputs result in known outputs when a machine model learns and trains
#....the datasets with the predictor values, using the model, considered appropriate for customers decision to cancel or retain
#....services.

#    THE DECISION FOR THE RANDOM FOREST CLASSIFIER MODEL AS THE APPROPRIATE ALGORITHM FOR BUILDING A PREDICTIVE MODEL

# 1. The categorical numerical 'churn' column values cleary and definitely support the use of a classification model, and 
#....specifically the Random Forest Algorithm model to determine a course action, for multiple decision occurrences like 
#....like voting to establish the most or majority decision.

# 2. The multiple split decision nodes of Random Forest Classifier model on the 'churn' question, to cancel services (YES)
#....or not to cancel (NO) by Telco customers, with a more reliable prediction of the outcome ahead of time, is important 
#....for the Retention Team to proactively contact the predicted customers and avoid a negative impact on the business with
#....a huge customer turnover and the revenue generation capacity of Telco business.

#....The Random Forest Classifier Algorithm is considered a better trade-off with Decision Tree Classifier Algorithm because 
#....it is an ensembling learning algorithm that builds on a collection of multiple decision trees with decision nodes, and 
#....combines the output to establish majority for the course of action. It is therefore a more robust and reliable algorithm,
#....and has a higher level of accuracy, unlike the Decision tree classifier algorithm that builds with a single decision tree. 

#....The multiple decision nodes give an edge with several decision trees voting and filtering instead of just a single decision,
#....based on the number of estimators.


# 3. The Retention Team's request to know which customers are likely to churn and cancel their subscriptions,in order to find a way 
#....to flag such customers, so they can be contacted proactively, can therefore be addressed, using the predicted outcome from the 
#....Random Forest Classifier model.
```
