#####  Analysis Of The DataSet From The Unique Values To Determine Columns To Drop, Features And Target Columns 

```python
# 1. The Dataset provides historical information about customer's profile to identify them, the services subscribed to for
#... billing, the service contract, monthly billing which translates into monthly payment of monthly charges and cummulates 
#... as total charges. A customer's profile and subscribed services is therefore the basis and determinig features for monthly 
#... billing and invariably monthly charges.

# 2. TO DROP CUSTOMER ID COLUMN

#... The customerID column is not a relevant or needed column because it is just an identifier of the customers and has no effect 
#... on historical revenue generated by customers as well as any current monthly charges 

# 3. CUSTOMER PROFILE AND SUBSCRIBED SERVICES COLUMNS FOR BILLING AND PAYMENTS AS FEATURES... REASONING AND JUSTIFICATION

# a. The columns for customer's profile and subscribed services apart from ID column are all useful as features columns and relevant
#... to predict the revenue estimates a customer is likely to generate on current basis and not on cummulative or historical revenue.
#... The profile abd billing features columns with int64/float64 are observed to have continuous values, indicating that a linear
#... regression model is chosen or decided to test and train the features columns for the prediction of likely revenue estimates that 
#... a Telco Customer can generate in a month.
# b. The binary features columns with Yes/No values need to be encoded, using map.apply to convert them to 0,1. 
# c. The other categorical features columns with a combination of mixed multiple texts and numerrical values, need to be encoded 
#... to 0,1 using One-HotEncoder, whiich is suitable and appropriate for features columns to make them readable for the model to learn 
#... and predict the likely revenue generation estimates of a customer as requested by the Finance Team.

# 4. MONTHLY CHARGES CONSIDERATION AND DECISION AS A MORE APPROPRIATE COLUMN TO USE AS TARGET FOR THE BUSINESS REQUEST. 
#... REASONING AND JUSTIFICATION

# a. The historical customer profile, 'tenure' and 'contract' records determine monthly revenue (monthly charges) based on
#... subscribed monthly services only. 
# b. 'Monthly charges' do not apply to all available services, but only to subscribed services in a month because 'tenure' and 'contract'
#... do not evenly apply or recur monthly for all customers. Monthly charges is therefore actual current charges. This is a more 
#... appropriate revenue measure to predict or estimate how much revenue a customer is likely to generate for sscenario 1.
# c. 'Monthly charges', based on current monthly billing is therefore considered the apprpriate customer 'revenue' genetated per month,
#... as the target column.

# 5. TO DROP TOTAL CHARGES COLUMN AND USE MONTHLY CHARGES COLUMN AS TARGET... REASONING AND JUSTIFICATION

# a..The 'total charges' (total revenue) on the other hand is historical and shows cummulative revenue generated. This includes 
#....various monthly services with different 'tenure', 'contract' months and type (month-to-month, 1-year or 2-years). 
# b..The data information about the 11 rows with blank spaces in 'total charges' column, due to 0 values on 'tenure; and 'senior citizens', 
#... is an indicatation the rows are for new customers and subscribers, who do not yet have historical billing and charges values, hence 0.
# c..The 'total charges' column drop is also considered necessary because it is closely related to both the 'tenure' and 'monthly charges' 
#....columns and the cummulative values indicated. 
# d..The column is also better dropped since, including it in the features columns could result in some data leakage in the model and cause
#....a misleading outcome, given that its relationship with 'tenure' is historical rather than the required likely estimate of how much 
#....'revenue' a customer can generate (current) as requested.
# e..All the considerations above, logically require that 'total charges' column be excluded and dropped from the predictive features columns 
#....required to determine the likely estimate of revenue generated by a customer and therefore not needed for the model. 

# 6. The basis for 'monthly charges' is just the sevices subscribed to in a month even though contact type may not be monthly and 
#....customers 'tenure' may also vary from month to month. Actual 'monthly charges' is therefore different from 'total charges'. 

# 7. TO DROP CHURN COLUMN... REASONING AND JUSTIFICATION

#... This column needs to be excluded because it is indicative of a customer's likely decision to stop subscribing for services, and a 
#... future outcome that is not a feature of current services subscribed to, for billing, revenue and payment. The churn indication and
#... situation invariably relates to a future outcome that has no effect whatsover on customer revenue generation estimates since the actual
#... churn information is not yet available for the customer revenue estimate.
```

#####  Analysis Of Obsevations On The Encoded Columns

```python
# Observations And Reasoning On The OneHotEncoder And Mapping Conversion Of Categorical Columns To Numeric

# 1. The encoding actions only focused on the remaining categorical features columns with the combined text and binary values.
# 2. Those categorical columns are with OneHotEncoder, applied only on features columns, converted to numeric. 
# 3. These columns are now named X_train_Encoded

# Observaytions Of The Mapping On The Binary Columns (yes/No) 

# 1. Only a few binary columns were converted to numeric.
# 2. Those binary-numeric columns were excluded from the OneHotEncoding action and effects.
# 3. These columns are still X-train

# Implications Of The Separate Actions

# 1. The features columns, though already converted to numeric, with separated and different applications of convertion 
#....methods are now X-train and X_train_encoded
# 2. This separate features (X) columns of categorical numeric-binary (X_train) and categorical numeric-multiple text/binary
#....(X_train_Encoded and need to be merged together to have a single features for training and testing of equal rows as 
#....Y_train and Y_test. 

#  Required Actions

# 1. To merge the entire categorical mapped binary features columns with the encoded text and binary features columns, all 
#....converted to numeric, and have a combined and single numeric features columns required to complete the data preprocessing 
#....and preparation for building a model to estimate the likely estimate of revenue a customer can generate for the Finance Team.
```

#####  Combining the Numeric Binary Columns And Numeric Categorical Encoded Columns

#####  1 - Converting The Encoded Values (Arrays) To DataFrames To Be Able To Combine Them With The Numeric And Binary_Mapped
#####      And Encoded Columns As A Single Features Columns

```python
X_train_Encoded = pd.DataFrame(X_train_Encoded, columns=Encode.get_feature_names_out(Categorical_Columns), index=X_train.index)
X_test_Encoded = pd.DataFrame(X_test_Encoded, columns=Encode.get_feature_names_out(Categorical_Columns), index=X_test.index)
```

##### 2 - Listing And Identifying The Numeric And Binary-Mapped Encoded Columns Still Reflected As X_Train

```python
X_train_Numeric = [Columns for Columns in X_train if Columns not in Categorical_Columns]
X_test_Numeric = [Columns for Columns in X_test if Columns not in Categorical_Columns]
```

##### 3 - Combining X_train (Numerical int64/float64 Columns And Numerical Binary-Mapped Columns) With The Encoded Multiple
#####     Categorical Text And Binary Values, To Have A Complete Single Features Columns For The Model For Both Training And
#####     Testing Sets Using pd.concat( ) As A Better Application Than Using Concat

```python
X_train_Combined = pd.concat([X_train[X_train_Numeric], X_train_Encoded], axis = 1)
```

```python
X_test_Combined = pd.concat([X_test[X_test_Numeric], X_test_Encoded], axis = 1)
```

##### Reasoning And Decision For The Choice Of Linear Regression Algorithm

```python
#    THE DATASET

# 1. The Telco Customers Dataset is a combination of different categories of features (X) data values with 21 columns, 
#....and a target column (Y).

# 2. The break-down of the different categories of the data values and their data types shows some of the feature columns 
#....with continuous numerical values, combined with other categorical columns with simpleas binary values and multiple
#....categories of texts, as well as simple and more complex binary values. 

# 3  The dataset are also labelled with inputs resulting in known outputs as the predictor columns for a supervised learning.

# 4  The labelled datasets show the relationship between inputs and outputs (independent i.e. features (X) and dependent i.e.
#....target (Y) values, for predicting a continuous values of target (Y) of labels of unseen and new data with a machine learning
#....algorithm, wherein inputs result in known outputs when a machine model learns and trains the datasets with the predictor
#....values, using the a regression model, appropriately considered to be linear regression algorithm.

#    THE DECISION FOR A LINEAR REGRESSION AS A MORE APPROPRIATE ALGORITHM FOR BUILDING A PREDICTIVE MODEL

# 1. The continuous numerical 'tenure' feature column values cleary and definitely support the use of Linear Regression model
#....in spite of the other categorical features columns.

# 2. The already known and/or  pre-determined values of the 'tenure' column from Telco customers profiles and subsribed services
#....provides the trade-off for the choice of linear regression algorithm over logistic regression algorithm.

# 3. The Finance Team request for the estimate of revenue that a customer is likely to generate, based on their profile and the 
#....services signed for can therefore be predicted with a linear regression model with the features columns identified and 
#....retained and the target column of Monthly Charges considered more appropriate to determine and predict a customer revenue
#....estimate than the historical and cumulative dropped Total Charges column which is more approriate for cumulative revenue
#....over previous periods and not current revenue estimate.
```
